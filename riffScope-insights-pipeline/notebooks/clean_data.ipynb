# The Revival of Alternative Music: A Response to the Exhaustion of Commercial Music?


## Project Overview

In recent years, alternative music has experienced renewed interest among listeners. This phenomenon has sparked discussions about whether this resurgence represents a reaction to the standardization of commercial music or a search for more authentic and emotionally complex artistic expressions.
The purpose of this project is to analyze the emotional and sonic characteristics of alternative music tracks in order to identify patterns that distinguish them from the dominant trends found in contemporary commercial music.
Through the analysis of musical variables obtained from APIs and other music-related data sources, this study explores whether factors such as emotional tone, intensity, acoustic production, and danceability reflect a growing preference for music perceived as more human, imperfect, and emotionally authentic.



## Research Question

**Do contemporary alternative music tracks exhibit emotional and sonic characteristics that differentiate them from dominant commercial music?**


## Variables Analyzed

| Variable | Description | Type |
|-----------|-------------|-------------|
| `valence` | Perceived emotional positivity or negativity of a track. | float64
| `energy` | Level of intensity and activity. |float64
| `acousticness` | Degree of acoustic production. |float64
| `danceability` | Suitability of a track for dancing. |float64
| `release_year` | Release date information. |datetime
| `name` | Track name. |string
| `artists` | Artist name. |string
| `id` | Unique track identifier. |string
| `group` | Category to which the track belongs (Alternative or Commercial). |string

## Objectives

- Compare the emotional characteristics of alternative and commercial music.
- Analyze differences in acoustic production and energy levels.
- Identify patterns associated with alternative music.
- Explore whether alternative music reflects a shift toward more emotionally complex and authentic artistic expressions.


## Expected Outcome

This project aims to identify whether alternative music demonstrates distinct emotional and sonic patterns when compared to commercial music, potentially reflecting changing listener preferences and broader cultural trends.

```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
from riffscope.features import read_csv,merge_df
```

```python
alternative_df,comercial_df = read_csv()
df_merge = merge_df(alternative_df,comercial_df)
df_merge.info()
df_merge.head()
```

```python

dups = df_merge[df_merge.duplicated(subset=['id'], keep=False)]
pd.set_option("display.max_rows",80)
dups[["id","name","artists","group"]].sort_values("id")
```

```python
dups_name = df_merge[df_merge.duplicated(subset=['name'], keep=False)]
pd.set_option("display.max_rows",200)
dups_name[["id","name","artists","group"]].sort_values("name")
```

```python
def clean_data(df):
    df = df.sort_values("group",ascending=True)
    df= df.drop_duplicates(subset=['id'],keep="first")
    df= df.drop_duplicates(subset=["name","artists"],keep="first")
    df["release_date"] = pd.to_datetime(df["release_date"],format="mixed")
    df["release_year"] = df["release_date"].dt.year
    df = df.drop(columns=["release_date"])
    df=df.reset_index(drop=True)
    return df
```

```python
data = clean_data(df_merge)
dup = data[data.duplicated(subset=["id"], keep=False)]
pd.set_option("display.max_rows",80)
dup[["id","name","artists","group"]].sort_values("id")
```
