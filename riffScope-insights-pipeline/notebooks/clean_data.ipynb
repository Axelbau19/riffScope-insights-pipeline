## Data Cleaning

Before modeling, the merged dataset required cleaning on three fronts: duplicate records, inconsistent date formats, and a redundant date column.

### Duplicate Discovery

Inspecting the categorical columns revealed that `id` and `name` had fewer unique values than expected for 1,785 rows, signaling duplicates.

| Column | Unique values |
|--------|--------------|
| `group` | 2 |
| `artists` | 957 |
| `id` | 1,745 |
| `name` | 1,679 |

Two types of duplicates were identified:

**1. Same track ID, different group label** — The same Spotify track appeared in both `alternative` and `comercial` because certain artists (e.g. Fleetwood Mac, Paramore, blink-182) were queried under both genre buckets.

**2. Same song name + artist, different Spotify ID** — Different releases or regional editions of the same track produced separate IDs (e.g. different pressings of *Iris* by The Goo Goo Dolls).

```python

dups = df_merge[df_merge.duplicated(subset=['id'], keep=False)]
pd.set_option("display.max_rows",80)
dups[["id","name","artists","group"]].sort_values("id")
```

```python
dups_name = df_merge[df_merge.duplicated(subset=['name'], keep=False)]
pd.set_option("display.max_rows",200)
dups_name[["id","name","artists","group"]].sort_values("name")
```

```python
def clean_data(df):
    df = df.sort_values("group",ascending=True)
    df= df.drop_duplicates(subset=['id'],keep="first")
    df= df.drop_duplicates(subset=["name","artists"],keep="first")
    df["release_date"] = pd.to_datetime(df["release_date"],format="mixed")
    df["release_year"] = df["release_date"].dt.year
    df = df.drop(columns=["release_date"])
    df=df.reset_index(drop=True)
    return df
```

### Cleaning Strategy

The `clean_data()` function resolves the issues above in one pass:

1. **Sort by `group` ascending** — puts `alternative` before `comercial`, so when duplicates are dropped the `alternative` label is kept (deterministic tie-breaking).
2. **Drop duplicates by `id`** — removes exact same tracks that appeared in both groups.
3. **Drop duplicates by `name` + `artists`** — removes same-song-different-ID cases.
4. **Parse `release_date` → `release_year`** — `release_date` had mixed formats (`YYYY-MM-DD` and `YYYY`); `pd.to_datetime(format="mixed")` handles both. Only the year is needed for analysis.
5. **Drop `release_date`** — the original column is no longer needed once the year is extracted.
6. **Reset index** — produces a clean 0-based index after all removals.

### Why Keep the `alternative` Label?

When a track appeared in **both groups**, a tie-breaking rule was needed. Sorting `group` in ascending alphabetical order places `"alternative"` before `"comercial"`, so `drop_duplicates(keep="first")` always retains the `alternative` label.

This choice is intentional and aligned with the research question. Artists flagged as duplicates — Fleetwood Mac, Paramore, blink-182, among others — originated in the alternative or rock space before crossing into the mainstream. Keeping them under `alternative` preserves their sonic and cultural identity, which is precisely what this study aims to compare against commercial music.

Classifying them as `comercial` would introduce noise into the alternative group in the wrong direction, potentially masking the very differences the analysis is designed to detect.

| Sorting order | Result |
|---|---|
| `"alternative"` (a…) | kept — appears first |
| `"comercial"` (c…) | dropped — appears second |

```python
data = clean_data(df_merge)
dup = data[data.duplicated(subset=["id"], keep=False)]
pd.set_option("display.max_rows",80)
dup[["id","name","artists","group"]].sort_values("id")
```

### Verification

After cleaning, deduplication by `id` returns an empty DataFrame — all exact-ID duplicates removed. 

Remaining apparent duplicates by `name` are **false positives**: coincidental song titles shared by completely different artists and genres (e.g. *Circles* by Pierce The Veil vs. Phatg; *Stay* by FISHER vs. Zedd). These are distinct tracks and are correctly retained.
